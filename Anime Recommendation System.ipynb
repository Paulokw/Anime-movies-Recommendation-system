```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\pc\Downloads\anime_with_synopsis.csv\anime_with_synopsis.csv")
df.head()
```

```python
df['comb']=df['Genres']+' '+df['sypnopsis']
```

```python
df=df.drop(columns=['Genres','sypnopsis'])
```

```python
df['comb']=df['comb'].fillna('').str.replace(r'[^a-zA-Z0-9]',' ', regex=True)
df.isna().sum()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(ngram_range=(2, 3))
```

```python
vec=tfidf.fit_transform(df['comb'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vec)
```

```python
idx=df[df['Name']=='Witch Hunter Robin'].index[0]
```

```python
distance=sorted(list(enumerate(similarity[idx])),reverse=True,key=lambda x:x[1])
```

```python
for i in distance[1:11]:
    print(df.iloc[i[0]].Name)
```

```python
def Recommend(anime_name):
    idx=df[df['Name']==anime_name].index[0]
    distance=sorted(list(enumerate(similarity[idx])),reverse=True,key=lambda x:x[1])
    for i in distance[1:11]:
        print(df.iloc[i[0]].Name)
```

```python
Recommend('Bouken Ou Beet')
```
